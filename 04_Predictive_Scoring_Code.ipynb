```python
import pandas as pd
import numpy as np
from pathlib import Path
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score, classification_report
import shap
import matplotlib.pyplot as plt
```

```python
# Download the original dataset separately; it is not redistributed here.
# Place Bank_Marketing.csv in the working directory or update DATA_PATH.
DATA_PATH = Path('Bank_Marketing.csv')
if not DATA_PATH.exists():
    raise FileNotFoundError(f'Place the downloaded dataset at {DATA_PATH.resolve()}')
df = pd.read_csv(DATA_PATH, sep=';')
if 'y' not in df.columns:
    raise ValueError('Expected a target column named y')
y = df['y'].map({'yes': 1, 'no': 0})
if y.isna().any():
    raise ValueError('Unexpected target values in y')
# Pre-contact modeling: duration is known only after a call, so exclude it.
X = df.drop(columns=['y', 'duration'], errors='ignore')
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)
cat_cols = X_train.select_dtypes(include=['object', 'category']).columns.tolist()
num_cols = [c for c in X_train.columns if c not in cat_cols]
preprocess = ColumnTransformer([
    ('categorical', OneHotEncoder(handle_unknown='ignore'), cat_cols),
    ('numeric', 'passthrough', num_cols),
])
model = Pipeline([
    ('preprocess', preprocess),
    ('classifier', RandomForestClassifier(n_estimators=300, max_depth=12,
                                           random_state=42, n_jobs=-1)),
])
model.fit(X_train, y_train)
y_prob = model.predict_proba(X_test)[:, 1]
print(f'Pre-contact holdout ROC-AUC: {roc_auc_score(y_test, y_prob):.4f}')
print(classification_report(y_test, model.predict(X_test), zero_division=0))
```

```python
# Explain the classifier on transformed holdout features.
# SHAP outputs differ by version; handle list, 2D and 3D arrays.
transformed = model.named_steps['preprocess'].transform(X_test.iloc[:80])
if hasattr(transformed, 'toarray'):
    transformed = transformed.toarray()
feature_names = model.named_steps['preprocess'].get_feature_names_out()
transformed_df = pd.DataFrame(transformed, columns=feature_names)
classifier = model.named_steps['classifier']
explainer = shap.TreeExplainer(classifier)
raw_sv = explainer.shap_values(transformed_df)
if isinstance(raw_sv, list):
    sv = np.asarray(raw_sv[1])
else:
    raw_sv = np.asarray(raw_sv)
    sv = raw_sv[:, :, 1] if raw_sv.ndim == 3 else raw_sv
base_values = np.asarray(explainer.expected_value)
base = float(base_values[1] if base_values.ndim else base_values)
shap.summary_plot(sv, transformed_df, show=False)
plt.title('Pre-contact model: SHAP summary')
plt.tight_layout()
plt.savefig('../images/05_SHAP_Summary.png', dpi=160, bbox_inches='tight')
plt.show()
```

```python
best_idx = int(np.argmax(y_prob[:len(transformed_df)]))
shap.plots.waterfall(shap.Explanation(
    values=sv[best_idx], base_values=base,
    data=transformed_df.iloc[best_idx].to_numpy(),
    feature_names=feature_names.tolist()), show=False)
plt.tight_layout()
plt.savefig('../images/06_waterfall.png', dpi=160, bbox_inches='tight')
plt.show()
```
